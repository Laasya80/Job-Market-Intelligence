# JMI — Notebook 06: Candidate Skill Gap Engine & Prioritization Algorithm

**Mathematical Formulation:**
$$\text{Match Score} = \frac{\sum_{\text{matched}} w_i + 0.5 \sum_{\text{partial}} w_j}{\sum_{\text{required}} w_k} \times 100$$
$$\text{Priority Score} = 0.50 \times \text{Market Demand} + 0.30 \times \text{Growth Rate} + 0.20 \times \text{Gap Weight}$$


```python
candidate_skills = {'python', 'sql', 'excel'}
required_skills = {'sql': 90, 'excel': 80, 'power bi': 75, 'python': 65, 'tableau': 50}

matched = [s for s in required_skills if s in candidate_skills]
missing = [s for s in required_skills if s not in candidate_skills]

earned = sum(required_skills[s] for s in matched)
total = sum(required_skills.values())
score = (earned / total) * 100
print(f'Profile Match Score: {score:.1f}%')
print('Missing Skills:', missing)
```
