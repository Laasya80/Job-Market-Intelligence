# JMI — Notebook 02: Exploratory Data Analysis (EDA)

**Purpose:**
Exploratory statistical and visual analysis of the job market: role distributions, city concentrations, salary distributions, and experience requirements.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
df = pd.read_csv('../data/processed/cleaned_jobs.csv')
print(f'Total Analyzed Jobs: {len(df)}')
```

## 1. Role Distribution Analysis


```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df, y='normalized_role', order=df['normalized_role'].value_counts().index, palette='Blues_r')
plt.title('Job Openings Distribution by Canonical Role')
plt.xlabel('Number of Postings')
plt.ylabel('Job Role')
plt.show()
```

## 2. Salary Distribution by City (Metro vs Non-Metro)


```python
plt.figure(figsize=(10, 4))
sns.boxplot(data=df, x='city', y='salary_max', palette='Set2')
plt.title('Salary (Max LPA) Distribution across Major Indian Cities')
plt.xlabel('City')
plt.ylabel('Salary (LPA INR)')
plt.show()
```
