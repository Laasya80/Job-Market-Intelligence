**Mathematical Formulation:**
$$\text{Emerging Score} = 0.50 \times \text{Norm Growth} + 0.30 \times \text{Norm Market Share} + 0.20 \times \text{Norm Volume}$$
$$\text{Growth Rate} = \frac{C_t - C_{t-1}}{\max(C_{t-1}, 1)} \times 100$$
