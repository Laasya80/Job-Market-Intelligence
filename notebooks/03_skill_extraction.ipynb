# JMI — Notebook 03: NLP & Taxonomy-Based Skill Extraction

**Purpose:**
Demonstrates multi-word phrase matching, word-boundary regex patterns, and category taxonomies to extract 100+ technologies from raw job descriptions.


```python
import re
import pandas as pd

TAXONOMY = {
    'Python': ['python', 'py', 'python3'],
    'SQL': ['sql', 'structured query language', 'postgresql', 'mysql'],
    'Machine Learning': ['machine learning', 'ml', 'statistical learning'],
    'Power BI': ['power bi', 'powerbi', 'dax'],
    'AWS': ['aws', 'amazon web services', 's3', 'ec2', 'lambda']
}

def extract_skills(text):
    matched = []
    for skill, aliases in TAXONOMY.items():
        for a in aliases:
            if re.search(rf'\b{re.escape(a)}\b', text, re.IGNORECASE):
                matched.append(skill)
                break
    return matched

sample_jd = 'Looking for an experienced Data Analyst skilled in Python, SQL querying, and building dashboards in Power BI.'
print('Extracted Skills:', extract_skills(sample_jd))
```
